```python
import pandas as pd

```

```python
df = pd.read_csv("Titanic-Dataset.csv")

df = df[(df["Age"] >= 18) & (df["Fare"] > 30)]
df
```

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["FarePerPerson"] = df["Fare"] / df["FamilySize"]
df
```

```python
analysis = df.groupby(["Pclass", "Sex"]).agg({
    "Fare": "mean",
    "Survived": "mean"
})
```
